```python
# Data preparation
df['date'] = pd.to_datetime(df['date'])

print(df['date'].head())
```

```python
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['is_weekend'] = df['date'].dt.weekday >= 5

df[['date', 'month', 'day', 'is_weekend']].head()
```

```python
import numpy as np
```

```python
df['unit'] = np.where(df['unit'] == 'Pack (100)', 'Box (100)', df['unit'])
df['unit'] = np.where(df['unit'] == 'Pack (50)', 'Box (50)', df['unit'])

df['unit'].value_counts()
```

```python
df.head(10)
```

```python
df['department_lag_1d_usage'] = (df.groupby(['product_id','department'])['quantity_used'].shift(1))

df['department_lag_1d_usage']
```

```python
df['department_lag_7d_usage'] = (df.groupby(['product_id','department'])['quantity_used'].shift(7))

df['department_lag_7d_usage']
```

```python
df['facility_stock_value_kes'] = (df['quantity_received']* df['unit_cost_kes'])

df['facility_stock_value_kes']
```
